# Vectors
---

Forces and moments of forces and torques are represented by vectors. A vector quantity has a magnitude and a unit. A vector has a starting point or point of application in three-dimensional space. The position of a point in space is specified with reference to the origin of a right-handed Cartesian coordinate system XYZ. The orientation of a vector in space can be specified with two angles, indicated here with $\theta$ and $\gamma$. 

Angle $\theta$ is the smallest angle between the projection of the vector on the XY-plane and an axis parallel to the X-axis having the same sense as the X-axis and passing through the starting point of the projected vector; the starting point of the projected vector can then be considered as the origin of the axis parallel to the X-axis. The angle is measured from the positive side of the parallel X-axis to the projected vector on the XY-plane. The absolute magnitude of the angle is limited between 0° and 180°. 

Angle $\gamma$ is the smallest angle measured from the vector's projection on the XY-plane to this vector. The absolute magnitude of the angle is limited between 0° and 90°. Both angles are considered to have a positive magnitude when turning in counter-clockwise direction and a negative magnitude in clockwise direction.

**Special cases**

|                                                                                           | $\theta$ | $\gamma$ |
|-------------------------------------------------------------------------------------------|:--------:|:--------:|  
| vector with starting point on the X-axis<br> and pointing in the same sense as the X-axis | 0°       | 0°       |
| vector with starting point on the Y-axis<br> and pointing in the same sense as the Y-axis | 90°      | 0°       |
| vector with starting point on the Z-axis<br> and pointing in the same sense as the Z-axis | 0°       | 90°      |

A vector in `python-mechanics` is represented by the `Vector` class (module `statics.vector.py`). From this class, class `Force` and class `Moment` are derived. When creating a `Force` or a `Moment` object, the position of the vector's starting point is specified through an object of class `Position`. The angles $\theta$ and $\gamma$ are objects of class `Angle`.

## Force

```python
from mechanics import Force
print_doc_string(Force.__init__)
```

> **Note**<br>
The docstring above is actually the docstring of the base class `Vector`. Derived class `Force` does not override the `__init__(...)` of class `Vector`, so it has inherited also the docstring of `Vector.__init__(...)`. However, do note that, while the defaults units of class `Vector` are dimensionless, the default units of class `Force` is newtons ('N'). So, if parameter `units` is left to default `None` when instantiating a `Force` object, the `Force` object will not be dimensionless; it will have units of 'N'. 

## Moment

The moment of a force or a torque is represented by an object of class `Moment`. Except for the measuring units, the parameters needed to create a `Moment` object are identical to the parameters that are needed to create a `Force` object. The default units (if parameter `units` is left to default `None`) of a `Moment` object are newton  meters ('N * m').
